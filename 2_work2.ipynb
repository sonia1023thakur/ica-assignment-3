```python
import pandas as pd
import numpy as np

# Load cleaned teaching dataset
df = pd.read_csv('teaching_prices_clean.csv')
df['date'] = pd.to_datetime(df['date'])
df = df.sort_values(['item', 'date', 'city']).reset_index(drop=True)
df.head(8)
```

## Part 1A: Summary Statistics & First-to-Last Percentage Change (Teaching Data)

```python
summary_rows = []

for item, group in df.groupby('item'):
    mean_price = group['price'].mean()
    median_price = group['price'].median()
    std_price = group['price'].std()
    
    # Weekly cross-city average price on first and last dates
    by_date = group.groupby('date')['price'].mean().sort_index()
    first_val = by_date.iloc[0]
    last_val = by_date.iloc[-1]
    pct_change_overall = ((last_val - first_val) / first_val) * 100
    
    # City-level first-to-last percentage changes
    c1 = group[group['city'] == 'City 1'].sort_values('date')
    c2 = group[group['city'] == 'City 2'].sort_values('date')
    pct_c1 = ((c1['price'].iloc[-1] - c1['price'].iloc[0]) / c1['price'].iloc[0]) * 100
    pct_c2 = ((c2['price'].iloc[-1] - c2['price'].iloc[0]) / c2['price'].iloc[0]) * 100
    
    summary_rows.append({
        'Item': item,
        'N': len(group),
        'Mean (PKR)': round(mean_price, 2),
        'Median (PKR)': round(median_price, 2),
        'Std Dev': round(std_price, 2),
        'First Avg (2026-01-01)': round(first_val, 2),
        'Last Avg (2026-03-19)': round(last_val, 2),
        'First-to-Last % Change (Overall)': round(pct_change_overall, 2),
        'City 1 % Change': round(pct_c1, 2),
        'City 2 % Change': round(pct_c2, 2)
    })

summary_df = pd.DataFrame(summary_rows)
print(summary_df.to_string(index=False))
```
